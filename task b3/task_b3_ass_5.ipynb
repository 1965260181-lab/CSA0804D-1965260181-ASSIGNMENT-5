```python
def mean_by_station_loops(
    rows: list[tuple[str, str, float]]
) -> dict[str, float]:

    sums, counts = {}, {}

    for st, _, pm in rows:
        sums[st] = sums.get(st, 0.0) + pm
        counts[st] = counts.get(st, 0) + 1

    return {
        st: round(sums[st] / counts[st], 2)
        for st in sums
    }
```
