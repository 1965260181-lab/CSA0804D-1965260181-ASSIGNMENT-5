```python
POSITIVE = {
    "good", "clean", "fast", "helpful", "improved"
}

NEGATIVE = {
    "bad", "dirty", "slow", "rude", "smoke", "broken"
}

NEGATORS = {
    "not", "never", "no"
}


def score(text: str) -> int:
    """+1 per positive word, -1 per negative word;
    a preceding negator flips it."""

    words = text.lower().replace(".", "").replace(",", "").split()

    total = 0

    for i, w in enumerate(words):
        val = 1 if w in POSITIVE else (-1 if w in NEGATIVE else 0)

        if i > 0 and words[i - 1] in NEGATORS:
            val = -val

        total += val

    return total


def label(text: str) -> str:
    s = score(text)

    return (
        "positive" if s > 0
        else "negative" if s < 0
        else "neutral"
    )


for c in [
    "Roads are clean and the bus was fast.",
    "Smoke from the factory, air is bad.",
    "The staff were not helpful.",
    "Not bad, service improved."
]:
    print(f"{label(c):8} | {c}")
```
