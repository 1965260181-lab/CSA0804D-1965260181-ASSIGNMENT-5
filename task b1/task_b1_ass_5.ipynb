```python
import csv
import io
import sqlite3

RAW = """station,date,pm25
Adyar,2026-10-01,48
Adyar,2026-10-02,NA
Adyar,2026-10-03,
Guindy,2026-10-01,95
Guindy,2026-10-02,-5
Guindy,2026-10-03,121"""


def load_clean(text: str) -> list[tuple[str, str, float]]:
    rows = []
    reader = csv.DictReader(io.StringIO(text.strip()))

    for r in reader:
        val_str = (r.get("pm25") or "").strip()

        if not val_str or val_str.upper() == "NA":
            continue

        try:
            pm = float(val_str)
        except ValueError:
            continue

        if 0.0 <= pm <= 999.0:
            rows.append(
                (r["station"].strip().title(), r["date"].strip(), pm)
            )

    return rows


def min_max(values: list[float]) -> list[float]:
    if not values:
        return []

    lo, hi = min(values), max(values)

    if hi == lo:
        return [0.0 for _ in values]

    return [round((v - lo) / (hi - lo), 3) for v in values]


clean = load_clean(RAW)

print(clean)
print(min_max([pm for _, _, pm in clean]))
```
